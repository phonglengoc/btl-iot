# Training & INT8 Quantization of the TinyML 1D-CNN Model for ESP32-S3 (`T04`, `T05`)

**Project:** Privacy-Preserving Device-Free Indoor Fall & Intrusion Detection System using Wi-Fi CSI & BLE

This notebook performs:
1. Loading the 64-subcarrier OFDM multipath Wi-Fi CSI dataset (`EMPTY`, `STATIONARY`, `WALKING`, `FALL`).
2. Causal signal preprocessing using a **Hampel filter ($2K+1=5$)** and **4th-order Butterworth bandpass filter ($0.2-15\,\text{Hz}$)**.
3. Training the `1D-CNN` model on **NVIDIA GeForce RTX 5070**, applying symmetric `INT8` quantization, and exporting C++ weights to `firmware/node2_rx_edge_ai/include/model_data.h`.

```
from pathlib import Path
from train_tinyml_1dcnn import train_and_export_tinyml

base = Path('.')
metrics = train_and_export_tinyml(
    dataset_dir=base / 'dataset',
    models_dir=base / 'models',
    artifacts_dir=base / 'artifacts',
    firmware_header_path=base.parent / 'firmware' / 'node2_rx_edge_ai' / 'include' / 'model_data.h',
    epochs=65,
)
print('INT8 Accuracy:', metrics['int8_accuracy'], 'Fall F1:', metrics['int8_fall_f1'])
```

## Training & INT8 Quantization Evaluation Results

- **Training Device:** `NVIDIA GeForce RTX 5070 Laptop GPU`
- **FP32 Test Accuracy:** `97.08%`
- **Post-Quantization INT8 Accuracy (`T04`):** `96.75%` (Accuracy Drop: `0.32%`)
- **Fall Class (`FALL`) F1-Score:** `94.87%`
- **Embedded INT8 Flash Weight Footprint on ESP32-S3:** `3392 bytes` (`~3.3 KB`)

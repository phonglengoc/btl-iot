# Huấn luyện & Lượng tử hóa INT8 Mô hình TinyML 1D-CNN cho ESP32-S3 (`T04`, `T05`)

**Đề tài:** Hệ thống Radar Không Camera (Wi-Fi CSI & BLE) Phát hiện Té ngã và Xâm nhập

Notebook này thực hiện:
1. Nạp bộ dữ liệu đa đường 64 sóng mang Wi-Fi CSI (`EMPTY`, `STATIONARY`, `WALKING`, `FALL`).
2. Tiền xử lý tín hiệu bằng bộ lọc nhân quả **Hampel ($2K+1=5$)** và **Butterworth bậc 4 ($0.2-15\,\text{Hz}$)**.
3. Huấn luyện mạng `1D-CNN` trên GPU **NVIDIA GeForce RTX 5070**, lượng tử hóa đối xứng `INT8` và xuất mảng C++ `firmware/node2_rx_edge_ai/include/model_data.h`.

```
from pathlib import Path
from train_tinyml_1dcnn import train_and_export_tinyml

base = Path('.')
metrics = train_and_export_tinyml(
    dataset_dir=base / 'dataset',
    models_dir=base / 'models',
    artifacts_dir=base / 'artifacts',
    firmware_header_path=base.parent / 'firmware' / 'node2_rx_edge_ai' / 'include' / 'model_data.h',
    epochs=65,
)
print('INT8 Accuracy:', metrics['int8_accuracy'], 'Fall F1:', metrics['int8_fall_f1'])
```

## Kết quả Thực nghiệm Huấn luyện & Lượng tử hóa INT8

- **Thiết bị huấn luyện:** `NVIDIA GeForce RTX 5070 Laptop GPU`
- **Độ chính xác FP32:** `97.08%`
- **Độ chính xác sau lượng tử hóa INT8 (`T04`):** `96.75%` (Độ sụt giảm: `0.32%`)
- **F1-Score lớp Té ngã (`FALL`):** `94.87%`
- **Dung lượng trọng số INT8 nhúng trên Flash ESP32-S3:** `3392 bytes` (`~3.3 KB`)
